## 03 · Choosing a dtype

Not every number needs the same precision — and `float32` covers almost everything here. Check with `.dtype`; convert with `.float()` or `.to(torch.float32)`.

<sub>Source: lecture slides 9–12</sub>

**`cast.py`** — predict the output, then run.

````python
labels = torch.tensor([0, 1, 2])
print(labels.dtype)         # torch.int64
floats = labels.float()     # convert to float32
print(floats.dtype)         # torch.float32
````

<details><summary>▸ Expected output</summary>

```
torch.int64
torch.float32
```
</details>

> ⚠️ **Watch out — Integers and decimals default differently**  
> `[1, 2, 3]` defaults to **int64**. `[1.0, 2.0]` defaults to **float32**. Mixing the two in an operation often produces a dtype-mismatch error.
>
> The fix: cast one explicitly with `.float()` or `.long()`.

## 05 · The quiet truth: memory is flat

The quiet truth: however many dimensions a tensor *looks* like it has, memory is **one flat row**.

<sub>Source: lecture slides 17–22</sub>

🧪 *Interactive lab in the app:* **What you see vs what’s really there** — try it there to see this idea move.

**`stride.py`** — predict the output, then run.

````python
m = torch.tensor([[1,2,3,4],
                  [5,6,7,8]])
print(m.shape)     # torch.Size([2, 4])
print(m.stride())  # (4, 1)
````

<details><summary>▸ Expected output</summary>

```
torch.Size([2, 4])
(4, 1)
```
</details>

> ⚠️ **Watch out — You won’t touch strides directly**  
> They exist so operations like transpose and view can rearrange a tensor **without copying its data** — fast and memory-efficient.
>
> But occasionally a “non-contiguous” tensor can’t be `view`-ed, only `reshape`-ed. That is Lecture 05.

## 06 · Counting elements and bytes

`numel()` × `element_size()` = what a tensor costs in memory. Use it to size a batch or a model before you run out of memory.

<sub>Source: lecture slides 23</sub>

**`bytes.py`** — predict the output, then run.

````python
t = torch.randn(3, 4)
print(t.numel())          # 12  (3 × 4)
print(t.element_size())   # 4 bytes (float32)
print(t.numel() * t.element_size(),
      "bytes")            # 48 bytes total
````

<details><summary>▸ Expected output</summary>

```
12
4
48 bytes
```
</details>

## The rule — worth remembering

**Five facts worth keeping**

1. A tensor is a grid of numbers with a shape — one integer per axis.
2. Every tensor carries three facts: shape, dtype, device. Print them first when debugging.
3. `ndim` counts the axes. `numel()` counts the total elements.
4. Memory is flat. The stride tells PyTorch how to navigate it — which is why some reshapes are free and others require a copy.
5. The default dtype is `float32` for decimals, `int64` for integers. Cast explicitly when mixing them.
